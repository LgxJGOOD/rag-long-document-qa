````python
# 加载所需的包（Packages）

import json                         # 处理 JSON 格式数据的标准包
import torch                        # PyTorch 是一个深度学习框架
from typing import List
from transformers import (          # Transformers
    AutoModelForCausalLM,           # 自动加载因果语言模型
    AutoTokenizer,                  # 自动加载对应的 tokenizer
    BitsAndBytesConfig,             # 用来设置量化模型的参数
    pipeline                        # 提供简单的模型推理接口
)
from tqdm.autonotebook import tqdm  # 显示进度条（适用于 Jupyter Notebook）
import jieba                        # 中文分词工具
from rank_bm25 import BM25Okapi     # BM25 搜索模型
from sentence_transformters import SentenceTransformer  # 向量模型
````

## 加载查询

````python
# 加载问题（queries），来自一个 JSON 文件

queries = []
with open("rag_questions.json", "r", encoding="utf-8") as fp:
    data = json.load(fp)  # 将整个 JSON 文件加载为 Python 列表/字典
    for item in data:
        queries.append(item['query']) # 提取每个字典中的 'query' 字段
````

````python
# 加载知识库（小说《回声》），并分段处理

chunks = []                   # 存放每段文字的列表
chunk_size = 512              # 每段文字的长度上限
stride = 256                  # 每段间的步长（改进：使用重叠增强检索）

# 从完整文本文件加载
with open("novel_echo.txt", "r", encoding="utf-8") as fp:
    passage = fp.read()         # 读取整本小说的文字内容

length = len(passage)
for i in range(0, length, stride):
    chunk = passage[i:i+chunk_size]  # 按照 chunk_size 切分文字段落
    chunks.append(chunk)
````

## 检索

### 加载检索模型

````python
# 初始化两种检索模型：BM25 和向量模型

# 创建 BM25 模型所需的分词数据
tokenized_chunks = [list(jieba.cut(chunk)) for chunk in chunks]
bm25 = BM25Okapi(tokenized_chunks)  # 创建 BM25 检索模型

# 加载向量模型（Embedding Model）
embedding_model = SentenceTransformer("intfloat/multilingual-e5-large")

# 对所有 chunks 计算向量
chunks_embeddings = embedding_model.encode(chunks, show_progress_bar=True)
````

### 检索函数

````python
# 定义 BM25 检索函数
def bm25_retrieve(query: str, chunks: List[str], bm25: BM25Okapi) -> List[str]:
    tokenized_query = list(jieba.cut(query))  # 对查询进行分词
    scores = bm25.get_scores(tokenized_query) # 计算每段文字的分数
    rank = sorted(zip(chunks, scores), key=lambda x: x[1], reverse=True)  # 按分数排序
    return [chunk for chunk, _ in rank]   # 返回排序后的段落

# 定义向量检索函数
def embedding_retrieve(query: str, chunks: List[str], chunks_embeddings: List[str], embedding_model: SentenceTransformer) -> List[str]:
    query_embeddings = embedding_model.encode(query, prompt_name="query")  # 获取查询的向量表示
    scores = list(embedding_model.similarity(query_embeddings, chunks_embeddings)[0])  # 计算相似度
    rank = sorted(zip(chunks, scores), key=lambda x: x[1], reverse=True)  # 排序
    return [chunk for chunk, _ in rank]

# 定义 Reciprocal Rank Fusion 函数，融合多种排序
def reciprocal_rank_fusion(*ranked_lists, k=60) -> List[str]:
    scores = {}
    for rl in ranked_lists:
        for rank, doc_id in enumerate(rl, start=1):  # rank 从 1 开始
            scores[doc_id] = scores.get(doc_id, 0.0) + 1.0 / (k + rank)

    fused = sorted(scores.items(), key=lambda x: (-x[1], x[0]))  # 排序并融合
    return [d for d, _ in fused]
````

## 加载大语言模型

````python
# 设置量化参数，减少内存使用
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,                    # 使用 4-bit 权重
    bnb_4bit_use_double_quant=True,       # 启用双重量化
    bnb_4bit_quant_type="nf4",            # 使用 nf4 量化类型
    bnb_4bit_compute_dtype=torch.bfloat16 # 使用 bfloat16 进行计算
)
# 加载语言模型
llm = AutoModelForCausalLM.from_pretrained(
    pretrained_model_name_or_path="Qwen/Qwen2.5-Coder-3B-Instruct",
    quantization_config=bnb_config,
    torch_dtype=torch.bfloat16,
    low_cpu_mem_usage=True,
)
# 加载对应的 tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    "Qwen/Qwen2.5-Coder-3B-Instruct",
)
# 使用 pipeline 包装模型推理接口
llm_pipe = pipeline(
    "text-generation",           # 任务为文本生成
    model=llm,                   # 使用的模型
    tokenizer=tokenizer,        # tokenizer
    max_new_tokens=256,         # 响应最大长度
)
````

### 定义提示词

````python
# 定义提示词（prompt），用来告诉模型角色与任务

SIMPLE_SYSTEM_PROMPT = """
请根据提问，直接提供精简并准确的解答，避免与问题无关的内容。
""".strip()

SYSTEM_PROMPT = """
请根据**问题**，从用户提供的**《回声》小说段落**找出与问题相关的内容，直接提供精简并准确的解答，避免与问题无关的内容。
""".strip()

USER_PROMPT = """
### 问题
{query}
### 《回声》小说段落
{relevant_chunks}
""".strip()
````

## 运行 RAG

- 每个 block 对应一个基准方法
- 只需执行一个 block 来生成预测文件

### Level1

````python
# 简单基准方法：直接将 query 给模型回答，不使用知识库

responses = []
for query in tqdm(queries, desc="Response Generation"):
    chats = [
        {'role': "system", 'content': SIMPLE_SYSTEM_PROMPT},
        {'role': "user", 'content': query},
    ]
    response = llm_pipe(chats)[0]['generated_text'][-1]['content'].strip()
    responses.append(response[:512])
````

### Leve2

````python
# 中等基准方法：只使用 BM25 找前 10 个段落，给模型参考

responses = []
for query in tqdm(queries, desc="Response Generation"):
    relevant_chunks = "\n\n".join(bm25_retrieve(query=query, chunks=chunks, bm25=bm25)[:10])
    chats = [
        {'role': "system", 'content': SYSTEM_PROMPT},
        {'role': "user", 'content': USER_PROMPT.format(query=query, relevant_chunks=relevant_chunks)},
    ]
    response = llm_pipe(chats)[0]['generated_text'][-1]['content'].strip()
    responses.append(response[:512])
````

### Level3

````python
# 强基准方法：结合 BM25 + 向量检索，使用 RRF 融合排序

responses = []
for query in tqdm(queries, desc="Response Generation"):
    relevant_chunks_bm25 = bm25_retrieve(query=query, chunks=chunks, bm25=bm25)
    relevant_chunks_embedding = embedding_retrieve(
        query=query,
        chunks=chunks,
        chunks_embeddings=chunks_embeddings,
        embedding_model=embedding_model
    )
    relevant_chunks = "\n\n".join(reciprocal_rank_fusion(relevant_chunks_bm25, relevant_chunks_embedding)[:10])

    chats = [
        {'role': "system", 'content': SYSTEM_PROMPT},
        {'role': "user", 'content': USER_PROMPT.format(query=query, relevant_chunks=relevant_chunks)},
    ]
    response = llm_pipe(chats)[0]['generated_text'][-1]['content'].strip()
    print(response)
    responses.append(response[:512])
````

````python
# 改进方法：优化检索策略
# 1. stride=256 增加chunk重叠，提升召回率
# 2. 取15个chunk作为上下文

responses = []
for query in tqdm(queries, desc="Response Generation"):
    relevant_chunks_bm25 = bm25_retrieve(query=query, chunks=chunks, bm25=bm25)
    relevant_chunks_embedding = embedding_retrieve(
        query=query,
        chunks=chunks,
        chunks_embeddings=chunks_embeddings,
        embedding_model=embedding_model
    )
    # 融合两个检索结果，取前15个chunk
    relevant_chunks = "\n\n".join(reciprocal_rank_fusion(relevant_chunks_bm25, relevant_chunks_embedding)[:15])

    chats = [
        {'role': "system", 'content': SYSTEM_PROMPT},
        {'role': "user", 'content': USER_PROMPT.format(query=query, relevant_chunks=relevant_chunks)},
    ]
    response = llm_pipe(chats)[0]['generated_text'][-1]['content'].strip()
    print(response)
    responses.append(response[:512])
````

## 创建预测文件

````python
# 输出响应结果为 JSON 文件，文件名使用学号

STUDENT_ID = "202320108026"
with open(f"hw2-result-{STUDENT_ID}.json", "w") as fp:
    json.dump(responses, fp, ensure_ascii=False, indent=2)
````
